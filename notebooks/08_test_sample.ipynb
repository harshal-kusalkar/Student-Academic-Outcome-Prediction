```python
X_test_path = Path("data/processed/X_test.csv")

if not X_test_path.exists():
    raise FileExistsError()
```

```python
df = pd.read_csv(X_test_path, sep=';')
df.head()
```

```python
import random 

SIZE = len(df)

samples = random.sample(range(SIZE), 10)
# print(samples)

sample_df = df.iloc[samples, :].copy()
sample_df
```

```python
TEST_SAMPLE_PATH = Path("data")

if not TEST_SAMPLE_PATH.exists():
    raise FileExistsError()

sample_df.to_csv(TEST_SAMPLE_PATH / "test_sample.csv", sep=';')
```
